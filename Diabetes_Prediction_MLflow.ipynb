```python
import pandas as pd
from sklearn.datasets import load_diabetes

# Muat dataset diabetes langsung dari Scikit-Learn
diabetes = load_diabetes()

# Buat DataFrame
df = pd.DataFrame(
    diabetes.data,
    columns=['AGE', 'SEX', 'BMI', 'BP', 'S1', 'S2', 'S3', 'S4', 'S5', 'S6']
)
df['Y'] = diabetes.target

# Tampilkan 5 baris pertama
df.head()
```

```python
# Buat kolom 'Risk' berbasis threshold 211.5 (Percentile 75)
df_clean = df.copy()
df_clean['Risk'] = (df_clean['Y'] > 211.5).astype(int)

# Tampilkan ringkasan statistik
df_clean.describe()
```

```python
import mlflow
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Split data (30% test, 70% train)
X, y = df_clean[['AGE','SEX','BMI','BP','S1','S2','S3','S4','S5','S6']].values, df_clean['Y'].values
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=0)

# Set eksperimen MLflow
mlflow.set_experiment("diabetes-regression")

with mlflow.start_run():
    mlflow.autolog()
    model_reg = LinearRegression()
    model_reg.fit(X_train, y_train)
    print("Model Regresi Berhasil Dilatih!")
```
